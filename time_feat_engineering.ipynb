```python
import pandas as pd

matches = pd.read_csv("data/gold_match.csv")
match_context = pd.read_csv("data/gold_match_context.csv")
```

```python
weekday = match_context[["weekday", "match_id"]]
matches = pd.merge(matches, weekday, on="match_id")
```

```python
matches["kickoff_time_local"] = pd.to_datetime(matches["kickoff_time_local"], format='%H:%M:%S')
matches["kickoff_time_local"] = pd.to_datetime(matches["kickoff_time_local"], format='%H:%M:%S').dt.hour
matches["is_18_hours"] = matches["kickoff_time_local"] == 18
```

```python
matches["is_sunday"] = matches["weekday"] == 6
matches["is_fri_hol"] = matches["weekday"].isin([4, 5, 6])
```
